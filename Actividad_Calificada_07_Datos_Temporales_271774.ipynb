### 1.1 Conversión (5 pts)

1. Muestra el tipo de dato de `pickup` y `dropoff`.
2. Convierte ambas columnas a tipo fecha con `pd.to_datetime()`, indicando el **formato explícito** que corresponde a los valores.
3. Verifica que la conversión no generó `NaT`: compara el número de faltantes antes y después.

```python
# Tipo de dato original (al leer el CSV, las fechas vienen como texto: object)
print('Tipos antes de convertir:')
print(taxis[['pickup', 'dropoff']].dtypes)

# Faltantes antes de la conversión (valores nulos en el texto original)
faltantes_antes = taxis[['pickup', 'dropoff']].isna().sum()

# Conversión con formato explícito.
taxis['pickup'] = pd.to_datetime(taxis['pickup'], format='%Y-%m-%d %H:%M:%S')
taxis['dropoff'] = pd.to_datetime(taxis['dropoff'], format='%Y-%m-%d %H:%M:%S')

# Faltantes DESPUÉS de la conversión
faltantes_despues = taxis[['pickup', 'dropoff']].isna().sum()

# Comparación antes vs después
comparacion_nat = pd.DataFrame({
    'faltantes_antes': faltantes_antes,
    'faltantes_despues': faltantes_despues
})
comparacion_nat['nuevos_NaT'] = comparacion_nat['faltantes_despues'] - comparacion_nat['faltantes_antes']
print('\nComparación de faltantes (de', len(taxis), 'filas):')
print(comparacion_nat)

print('\nTipos después de convertir:')
print(taxis[['pickup', 'dropoff']].dtypes)
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
taxis['hora'] = taxis['pickup'].dt.hour
taxis['dia_num'] = taxis['pickup'].dt.dayofweek          # lunes = 0, domingo = 6
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()
taxis['fin_semana'] = taxis['dia_num'] >= 5              # sábado (5) o domingo (6)
taxis['fecha'] = taxis['pickup'].dt.date
taxis['cruza_medianoche'] = taxis['dropoff'].dt.date != taxis['pickup'].dt.date

print('Viajes que cruzan la medianoche:', taxis['cruza_medianoche'].sum())
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
por_hora = taxis.groupby('hora').agg(
    viajes=('hora', 'size'),
    mediana_duracion_min=('duracion_min', 'median')
)
print(por_hora)

hora_max = por_hora['viajes'].idxmax()
hora_min = por_hora['viajes'].idxmin()

print(f"\nHora con más viajes  : {hora_max}:00 ({por_hora.loc[hora_max, 'viajes']} viajes)")
print(f"Hora con menos viajes: {hora_min}:00 ({por_hora.loc[hora_min, 'viajes']} viajes)")
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
por_dia = taxis.groupby(['dia_num', 'dia_nombre']).agg(
    viajes=('dia_num', 'size'),
    mediana_duracion_min=('duracion_min', 'median'),
    mediana_fare=('fare', 'median'),
    fechas_distintas=('fecha', 'nunique')
)

por_dia['viajes_por_fecha'] = (por_dia['viajes'] / por_dia['fechas_distintas']).round(1)

print(por_dia)
```

## Actividad 4. Dataset limpio y comparación (20 puntos)

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
# Condiciones de exclusión de acuerdo a 2.4
c_periodo = ~((taxis['pickup'] >= '2019-03-01') & (taxis['pickup'] < '2019-04-01'))  # fuera de marzo 2019
c_corta   = taxis['duracion_min'] < 1      # duración < 1 min
c_larga   = taxis['duracion_min'] > 180    # duración > 3 h
c_veloz   = taxis['velocidad_mph'] > 60    # velocidad poco probable

print('Fuera de marzo 2019 :', c_periodo.sum())
print('Duración < 1 min    :', c_corta.sum())
print('Duración > 3 h      :', c_larga.sum())
print('Velocidad > 60 mph  :', c_veloz.sum())

# Un viaje puede cumplir varias condiciones
excluir = c_periodo | c_corta | c_larga | c_veloz
print('\nTotal excluidos (sin doble conteo):', excluir.sum())

# Dataset limpio: se conservan los que no cumplen ninguna condición
taxis_limpio = taxis[~excluir].copy()
print('Antes:', len(taxis), '| Después:', len(taxis_limpio))

# Verificación
assert taxis_limpio['pickup'].between('2019-03-01', '2019-03-31 23:59:59').all()
assert (taxis_limpio['duracion_min'] >= 1).all()
assert (taxis_limpio['duracion_min'] <= 180).all()
assert (taxis_limpio['velocidad_mph'] <= 60).all()
print('Verificación superada: taxis_limpio cumple todas las condiciones.')
print("Duración mínima:", taxis_limpio["duracion_min"].min(), "minutos")
print("Duración máxima:", taxis_limpio["duracion_min"].max(), "minutos")
print("Velocidad máxima:", taxis_limpio["velocidad_mph"].max(), "mph")
```
