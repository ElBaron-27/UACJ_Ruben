## Actividad 4 — Tipos de dato (30 pts)

```python
df['TotalCharges'].dtype
```

**4.1 — Investiga (responde en Markdown):**

`TotalCharges` contiene valores numéricos (montos en dólares), pero pandas la cargó como `object`, no como `float`. Investiga por qué — revisa si hay algún valor que no se vea como un número normal.

_Tu respuesta:_ Aunque se observan que la mayoría de los valores son numéricos se encontraron 11 registros que contienen un espacio en blanco (' ') en lugar de un número. Algo importantes es que al revisar estos registros, también se observa que corresponden a clientes con tenure = 0, por lo que probablemente todavía no cuentan con cargos totales acumulados. En conclusión, el problema no se encuentra en los valores numéricos de la columna, sino en esos valores vacíos almacenados como texto.

```python
# 4.2 — Corrige el tipo de TotalCharges.
# Pista: pd.to_numeric() con el parámetro errors= te puede ayudar a identificar
# o manejar los valores problemáticos que encontraste en 4.1.

# Identificamos los valores vacíos de TotalCharges
totalcharges_vacios = df['TotalCharges'].str.strip() == ''

print(f"Valores vacíos encontrados: {totalcharges_vacios.sum()}")

# Los registros vacíos corresponden a clientes con tenure = 0,
# por lo que sustituimos TotalCharges por 0 sin eliminar las filas
df.loc[totalcharges_vacios, 'TotalCharges'] = '0'

# Convertimos finalmente la columna a tipo numérico
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='raise')

# Verificamos el resultado
print(df['TotalCharges'].dtype)
print(f"Valores nulos: {df['TotalCharges'].isna().sum()}")
```

```python
# 4.3 — Convierte a category las columnas categóricas que, según lo que calculaste
# en la Actividad 3, tengan cardinalidad baja y valores fijos.
# Verifica con .dtypes que el cambio se aplicó correctamente.

columnas_category = [
    'gender',
    'SeniorCitizen',
    'Partner',
    'Dependents',
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies',
    'Contract',
    'PaperlessBilling',
    'PaymentMethod',
    'Churn'
]


for col in columnas_category:
    df[col] = df[col].astype('category')

print(df[columnas_category].dtypes)
```
